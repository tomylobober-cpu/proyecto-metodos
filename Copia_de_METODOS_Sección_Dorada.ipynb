#Busqueda de la sección dorada

```python
import numpy as np

def f(x):
  return 2*np.sin(x) - (x**2)/10
```

```python
x0 = 0
x3 = 4

R = ((5**0.5)-1)/2
d = (x3-x0)*R

x1 = x3 - d
x2 = x0 + d

#Ahora empezamos a interar, evaluando el error. Estoy mirando la altura de y(xi)

for i in range(10):
  if f(x1) > f(x2):
    x3 = x2
    x2 = x1
    d = (x3-x0)*R
    x1 = x3 - d
    val_opt = f(x1)
  else:
    x0 = x1
    x1 = x2
    d = (x3 - x0)*R
    x2 = x0 + d
    val_opt = f(x2)
  print(val_opt)
```

```python
x0 = 0
x3 = 4

R = ((5**0.5)-1)/2
d = (x3-x0)*R

x1 = x3 - d
x2 = x0 + d

n = 5 # de cifras significativas
error_deseado = 0.5 * 10**(-n)
opt_ant = 0
opt_act = x1
error = (opt_act - opt_ant)/opt_act
i = 0 #iteraciones

#Ahora empezamos a interar, evaluando el error. Estoy mirando la altura de y(xi)
while (error > error_deseado):
  i += 1
  opt_ant = opt_act
  if f(x1) > f(x2):
    x3 = x2
    x2 = x1
    d = (x3-x0)*R
    x1 = x3 - d
    opt_act = f(x1)
  else:
    x0 = x1
    x1 = x2
    d = (x3 - x0)*R
    x2 = x0 + d
    opt_act = f(x2)
  error = np.abs((opt_act - opt_ant)/opt_act)
  print(f"Iteración {i}")
  print(f"Optimo anterior: {opt_ant}, Optimo actual: {opt_act}, Error: {error}")
```

```python
import numpy as np

R = ((5**0.5)-1)/2
n = 5
error_deseado = 0.5 * 10**(-n)

# Punto de inicio
x_opt = 0
y_opt = 0

error_multivariado = 1
ciclos = 0

# --- BUCLE EXTERNO (Problema Multivariado) ---
while(error_multivariado > error_deseado):
    print("..................................................")
    ciclos += 1
    print(f"Inicio de Ciclo Multivariado {ciclos}")

    # Guardamos el punto con el que iniciamos este ciclo para el error multivariado
    x_ciclo_ant = x_opt
    y_ciclo_ant = y_opt

    # --- BUCLE INTERNO 1: PROBLEMA UNIVARIADO X ---
    x0, x3 = 0, 4
    d = (x3 - x0)*R
    x1 = x3 - d
    x2 = x0 + d

    error_x = 1
    while(error_x > error_deseado):
        x_opt_ant = x_opt

        if t(x1, y_opt) > t(x2, y_opt): # Manteniendo 'y_opt' como constante
            x3 = x2
            x2 = x1
            d = (x3 - x0)*R
            x1 = x3 - d
            x_opt = x1
        else:
            x0 = x1
            x1 = x2
            d = (x3 - x0)*R
            x2 = x0 + d
            x_opt = x2

        # Error Univariado X
        error_x = np.abs((x_opt - x_opt_ant)/x_opt) if x_opt != 0 else np.abs(x_opt - x_opt_ant)

    print(f"  -> X optimizado: {x_opt}")

    # --- BUCLE INTERNO 2: PROBLEMA UNIVARIADO Y ---
    y0, y3 = 0, 2
    d = (y3 - y0)*R
    y1 = y3 - d
    y2 = y0 + d

    error_y = 1
    while(error_y > error_deseado):
        y_opt_ant = y_opt

        if t(x_opt, y1) > t(x_opt, y2): # Manteniendo el nuevo 'x_opt' como constante
            y3 = y2
            y2 = y1
            d = (y3 - y0)*R
            y1 = y3 - d
            y_opt = y1
        else:
            y0 = y1
            y1 = y2
            d = (y3 - y0)*R
            y2 = y0 + d
            y_opt = y2

        # Error Univariado Y
        error_y = np.abs((y_opt - y_opt_ant)/y_opt) if y_opt != 0 else np.abs(y_opt - y_opt_ant)

    print(f"  -> Y optimizado: {y_opt}")

    # --- CÁLCULO DEL ERROR MULTIVARIADO ---
    # Se compara el (x,y) obtenido tras ambos bucles con el (x,y) del inicio del ciclo
    err_multi_x = np.abs((x_opt - x_ciclo_ant)/x_opt) if x_opt != 0 else np.abs(x_opt - x_ciclo_ant)
    err_multi_y = np.abs((y_opt - y_ciclo_ant)/y_opt) if y_opt != 0 else np.abs(y_opt - y_ciclo_ant)

    # El error multivariado es el mayor cambio relativo entre las dos variables
    error_multivariado = max(err_multi_x, err_multi_y)

    print(f"Error Multivariado actual: {error_multivariado}")

print(f"\nÓptimo final -> x: {x_opt}, y: {y_opt}")
```
