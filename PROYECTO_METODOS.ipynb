Interpolación directa

```python
x = [3.0, 4.0, 5.0]
y = [4.0, 5.0, 3.0]

A = []

for i in range(3):
    fila = []
    for j in range(3):
        fila.append(x[i]**(2-j))
    A.append(fila)
print(A)

import numpy as np
import matplotlib.pyplot as plt

coef = np.linalg.solve(A, y)
print(coef)

x_i = np.linspace(3,5,40)
y_i = []

for xs in x_i:
    y_eval = coef[0]*xs**2 + coef[1]*xs + coef[2]
    y_i.append(y_eval)

plt.plot(x_i, y_i)
plt.scatter(x,y)
```

Regresión directa

```python
import numpy as np
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7]
y = [0.5, 2.5, 2.0, 4, 3.5, 6, 5.5]
n = len(x)

A = []
for i in range(1, -1, -1):
    fila = []
    for j in range(1, -1, -1):
        fila.append(sum([val**(i + j) for val in x]))
    A.append(fila)

B = []
for i in range(1, -1, -1):
    if i == 1:
        B.append(sum(xv * yv for xv, yv in zip(x, y)))
    else:
        B.append(sum(y))

coef = np.linalg.solve(A, B)

x_i = np.linspace(1, 7, 100)
y_i = []
for x_s in x_i:
    y_i.append(coef[0] * x_s + coef[1])
```

Metodo por trapecios


```python
"""#Metodo de integracion por trapecios"""

import numpy as np

# 1. Funcion a integrar
def f(x):
    return x - np.cos(x)

# 2. Parametros de entrada
a = 0          # Limite inferior de integracion
b = 3          # Limite superior de integracion
n = 10         # Numero de trapecios (subintervalos)

# 3. Ancho de cada trapecio (paso h)
h = (b - a) / n

# 4. Calculo iterativo (sumando el area de cada trapecio individual)
# Formula del area de un trapecio: h * (f(x_i) + f(x_{i+1})) / 2
integral_aprox = 0
LISTA_TRAPECIOS = []

for i in range(n):
    x_actual = a + i * h
    x_siguiente = x_actual + h

    # Area del trapecio en el subintervalo [x_actual, x_siguiente]
    area_i = h * (f(x_actual) + f(x_siguiente)) / 2

    integral_aprox = integral_aprox + area_i
    LISTA_TRAPECIOS.append(area_i)

# 5. Valor exacto analitico para calcular el error (integral de x - cos(x) es x^2/2 - sin(x))
def integral_exacta(x):
    return (x**2) / 2 - np.sin(x)

valor_real = integral_exacta(b) - integral_exacta(a)
error_relativo = abs((valor_real - integral_aprox) / valor_real) * 100

print("El valor aproximado de la integral es:")
print(integral_aprox)
print("El valor real es:")
print(valor_real)
print("El error relativo porcentual es:")
print(error_relativo)
print("Areas de cada trapecio:")
print(LISTA_TRAPECIOS)
```

Metodo Simpson 1/3

```python
"""# Metodo de integracion por Simpson 1/3 compuesto"""

import numpy as np

# 1. Funcion a integrar
def f(x):
    return x - np.cos(x)

# 2. Parametros de entrada
a = 0          # Limite inferior de integracion
b = 3          # Limite superior de integracion
n = 10         # Numero de subintervalos (¡Debe ser PAR para Simpson 1/3!)

# Validacion para asegurar que n sea par
if n % 2 != 0:
    n += 1
    print(f"Aviso: n ajustado a {n} para cumplir con la condicion de ser par.")

# 3. Ancho del subintervalo (paso h)
h = (b - a) / n

# 4. Aplicacion de la formula compuesta de Simpson 1/3
# I = (h/3) * [f(a) + 4 * sum(terminos impares) + 2 * sum(terminos pares) + f(b)]
suma_impares = 0
suma_pares = 0

for i in range(1, n):
    x_i = a + i * h
    if i % 2 == 1:
        suma_impares = suma_impares + f(x_i)
    else:
        suma_pares = suma_pares + f(x_i)

integral_aprox = (h / 3) * (f(a) + 4 * suma_impares + 2 * suma_pares + f(b))

# 5. Valor exacto analitico para calcular el error (integral de x - cos(x))
def integral_exacta(x):
    return (x**2) / 2 - np.sin(x)

valor_real = integral_exacta(b) - integral_exacta(a)
error_relativo = abs((valor_real - integral_aprox) / valor_real) * 100

print("El valor aproximado de la integral por Simpson 1/3 es:")
print(integral_aprox)
print("El valor real es:")
print(valor_real)
print("El error relativo porcentual es:")
print(error_relativo)
print("Numero de subintervalos (n) utilizados:")
print(n)
```

Método Simpson 3/8

```python
"""# Metodo de integracion por Simpson 3/8 compuesto"""

import numpy as np

# 1. Funcion a integrar
def f(x):
    return x - np.cos(x)

# 2. Parametros de entrada
a = 0          # Limite inferior de integracion
b = 3          # Limite superior de integracion
n = 9          # Numero de subintervalos (¡Debe ser multiplo de 3 para Simpson 3/8!)

# Validacion para asegurar que n sea multiplo de 3
if n % 3 != 0:
    n = n + (3 - (n % 3))
    print(f"Aviso: n ajustado a {n} para ser multiplo de 3.")

# 3. Ancho del subintervalo (paso h)
h = (b - a) / n

# 4. Aplicacion de la formula compuesta de Simpson 3/8
# I = (3*h / 8) * [f(a) + 3 * sum(indices no divisibles por 3) + 2 * sum(indices divisibles por 3) + f(b)]
suma = f(a) + f(b)

for i in range(1, n):
    x_i = a + i * h
    if i % 3 == 0:
        suma = suma + 2 * f(x_i)
    else:
        suma = suma + 3 * f(x_i)

integral_aprox = (3 * h / 8) * suma

# 5. Valor exacto analitico para calcular el error
def integral_exacta(x):
    return (x**2) / 2 - np.sin(x)

valor_real = integral_exacta(b) - integral_exacta(a)
error_relativo = abs((valor_real - integral_aprox) / valor_real) * 100

print("El valor aproximado de la integral por Simpson 3/8 es:")
print(integral_aprox)
print("El valor real es:")
print(valor_real)
print("El error relativo porcentual es:")
print(error_relativo)
print("Numero de subintervalos (n) utilizados:")
print(n)
```
