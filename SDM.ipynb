```python
import numpy as np

def t(x, y):
  return (x-3)**2 + (y-5)**2 + 10

R = ((5**0.5)-1)/2
n = 5
error_deseado = 0.5 * 10**(-n)

# Punto de inicio
x_opt = 0
y_opt = 0

error_multivariado = 1
ciclos = 0

# --- BUCLE EXTERNO (Problema Multivariado) ---
while(error_multivariado > error_deseado):
    print("..................................................")
    ciclos += 1
    print(f"Inicio de Ciclo Multivariado {ciclos}")

    # Guardamos el punto con el que iniciamos este ciclo para el error multivariado
    x_ciclo_ant = x_opt
    y_ciclo_ant = y_opt

    # --- BUCLE INTERNO 1: PROBLEMA UNIVARIADO X ---
    x0, x3 = 0, 6
    d = (x3 - x0)*R
    x1 = x3 - d
    x2 = x0 + d

    error_x = 1
    while(error_x > error_deseado):
        x_opt_ant = x_opt

        if t(x1, y_opt) < t(x2, y_opt): # MINIMIZAR, manteniendo 'y_opt' constante
            x3 = x2
            x2 = x1
            d = (x3 - x0)*R
            x1 = x3 - d
            x_opt = x1
        else:
            x0 = x1
            x1 = x2
            d = (x3 - x0)*R
            x2 = x0 + d
            x_opt = x2

        # Error Univariado X
        error_x = np.abs((x_opt - x_opt_ant)/x_opt) if x_opt != 0 else np.abs(x_opt - x_opt_ant)

    print(f"  -> X optimizado: {x_opt}")

    # --- BUCLE INTERNO 2: PROBLEMA UNIVARIADO Y ---
    y0, y3 = 0, 10
    d = (y3 - y0)*R
    y1 = y3 - d
    y2 = y0 + d

    error_y = 1
    while(error_y > error_deseado):
        y_opt_ant = y_opt

        if t(x_opt, y1) < t(x_opt, y2): # MINIMIZAR, con el nuevo 'x_opt' constante
            y3 = y2
            y2 = y1
            d = (y3 - y0)*R
            y1 = y3 - d
            y_opt = y1
        else:
            y0 = y1
            y1 = y2
            d = (y3 - y0)*R
            y2 = y0 + d
            y_opt = y2

        # Error Univariado Y
        error_y = np.abs((y_opt - y_opt_ant)/y_opt) if y_opt != 0 else np.abs(y_opt - y_opt_ant)

    print(f"  -> Y optimizado: {y_opt}")

    # --- CÁLCULO DEL ERROR MULTIVARIADO ---
    err_multi_x = np.abs((x_opt - x_ciclo_ant)/x_opt) if x_opt != 0 else np.abs(x_opt - x_ciclo_ant)
    err_multi_y = np.abs((y_opt - y_ciclo_ant)/y_opt) if y_opt != 0 else np.abs(y_opt - y_ciclo_ant)

    # El error multivariado es el mayor cambio relativo entre las dos variables
    error_multivariado = max(err_multi_x, err_multi_y)

    print(f"Error Multivariado actual: {error_multivariado}")

print(f"\nÓptimo final -> x: {x_opt}, y: {y_opt}, z: {t(x_opt, y_opt)}")
```
